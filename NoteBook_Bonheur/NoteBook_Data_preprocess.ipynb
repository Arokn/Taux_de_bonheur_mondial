Data  set for "Acces eau potable"

```python
# Read the file without the metadata line 
import pandas as pd

df_water = pd.read_csv(
    "Initial_DataSet\Acces_eau_potable_PI_SH.H2O.BASW.ZS_DS2_en_csv_v2_4511.csv",
    skiprows=4
)
#Transform  the year to one column
df_water = df_water.melt(
    id_vars=["Country Name", "Country Code", "Indicator Name", "Indicator Code"],
    var_name="Year",
    value_name="Water_Access"
)
#Rename de columns 
df_water = df_water.rename(columns={
    "Country Name": "Country"
})
#Keep just columns "Country","Year", "Water_Access"
df_water = df_water[["Country", "Year", "Water_Access"]]
# Keep from 2011 to 2025
    #Convert from string to int for the comparaison after 
df_water["Year"] = pd.to_numeric(df_water["Year"], errors="coerce")
df_water["Water_Access"] = pd.to_numeric(df_water["Water_Access"], errors="coerce")

df_water = df_water[
    (df_water["Year"] >= 2011) & (df_water["Year"] < 2025)
].reset_index(drop=True)


print(df_water)
```
